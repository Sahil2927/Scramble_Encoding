```python
def b8zs_encode(bitstream):
 
    result = []
    last_polarity = -1   # So first '1' becomes +1
    i = 0
    n = len(bitstream)

    while i < n:
        # Check for 8 consecutive zeros
        if bitstream[i:i+8] == "00000000":
            
            if last_polarity == -1:
                # Last pulse was negative
                pattern = [0, 0, 0, -1, +1, 0, +1, -1]
            else:
                # Last pulse was positive
                pattern = [0, 0, 0, +1, -1, 0, -1, +1]

            result.extend(pattern)
            last_polarity = pattern[-1]
            i += 8
        
        else:
            if bitstream[i] == '1':
                last_polarity *= -1
                result.append(last_polarity)
            else:
                result.append(0)
            i += 1

    return result
```

```python
data = "1111000000000111"

encoded = b8zs_encode(data)

print("Input Bitstream: ", data)
print("B8ZS Encoded:    ", encoded)
```
