```python
def hdb3_encode(bitstream):
    result = []
    last_polarity = -1  # So first '1' becomes +1
    pulse_count = 0    # Count non-zero pulses since last violation
    i = 0
    n = len(bitstream)

    while i < n:
        # Check for 4 consecutive zeros
        if bitstream[i:i+4] == "0000":
            
            if pulse_count % 2 == 0:
                # EVEN → B00V
                B = -last_polarity
                V = B  # violation (same polarity)
                pattern = [B, 0, 0, V]
                last_polarity = V
            else:
                # ODD → 000V
                V = last_polarity  # violation
                pattern = [0, 0, 0, V]
                last_polarity = V

            result.extend(pattern)
            pulse_count = 0
            i += 4
        
        else:
            if bitstream[i] == '1':
                last_polarity *= -1
                result.append(last_polarity)
                pulse_count += 1
            else:
                result.append(0)
            i += 1

    return result
```

```python
data = "1111000000000111"

encoded = hdb3_encode(data)

print("Input:   ", data)
print("Encoded: ", encoded)
```
