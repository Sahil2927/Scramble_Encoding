```python
import numpy as np
```

```python
data = "1111000000000111"
key  = "1010101010101010"

# Convert to NumPy integer arrays
data_bits = np.array(list(data), dtype=int)
key_bits  = np.array(list(key), dtype=int)

print("Data bits: ", data_bits)
print("Key bits : ", key_bits)
```

```python
encoded_bits = np.bitwise_xor(data_bits, key_bits)

print("Encoded bits:", encoded_bits)
```

```python
encoded_string = ''.join(encoded_bits.astype(str))

print("Encoded Output:", encoded_string)
```

```python
re_encoded_bits = np.bitwise_xor(encoded_bits,key_bits)

print ("Re_encoded bits:", re_encoded_bits)
```

```python
re_encoded_string = ''.join(re_encoded_bits.astype(str))

print("Re_encoded Output:", re_encoded_string)
```
